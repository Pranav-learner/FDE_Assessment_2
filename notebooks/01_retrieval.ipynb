## 4B — Retrieval Validation

### 4B.1 Required Trip Columns

The retrieved trip dataset must contain the fields required to answer
the defined business questions and connect trip records to the taxi
zone reference data.

```python
required_trip_columns = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime",
    "PULocationID",
    "DOLocationID",
    "trip_distance"
]

missing_trip_columns = [
    col for col in required_trip_columns
    if col not in trips.columns
]

print("Required columns:", required_trip_columns)
print("Missing columns:", missing_trip_columns)

if missing_trip_columns:
    print("❌ Retrieval validation failed")
else:
    print("✅ All required trip columns are present")
```

```python
reporting_start = pd.Timestamp("2026-07-01 00:00:00")
reporting_end = pd.Timestamp("2026-07-31 23:59:59")

july_pickups = july_2026_trips["tpep_pickup_datetime"]

print("Reporting start:", reporting_start)
print("Reporting end:", reporting_end)

print("Actual minimum pickup:", july_pickups.min())
print("Actual maximum pickup:", july_pickups.max())

print(
    "All pickups within reporting period:",
    july_pickups.between(reporting_start, reporting_end).all()
)
```

```python
print("Zone rows:", len(zones))
print("Zone columns:")
print(zones.columns.tolist())

print("\nDuplicate LocationIDs:")
print(zones["LocationID"].duplicated().sum())

print("\nMissing LocationIDs:")
print(zones["LocationID"].isna().sum())
```

```python
zone_ids = set(zones["LocationID"].dropna())

pickup_location_ids = set(
    july_2026_trips["PULocationID"].dropna()
)

dropoff_location_ids = set(
    july_2026_trips["DOLocationID"].dropna()
)

unmatched_pickup_ids = pickup_location_ids - zone_ids
unmatched_dropoff_ids = dropoff_location_ids - zone_ids

print("Unique pickup location IDs:", len(pickup_location_ids))
print("Unique dropoff location IDs:", len(dropoff_location_ids))

print("Unmatched pickup IDs:", len(unmatched_pickup_ids))
print("Unmatched dropoff IDs:", len(unmatched_dropoff_ids))

print("Sample unmatched pickup IDs:", sorted(unmatched_pickup_ids)[:20])
print("Sample unmatched dropoff IDs:", sorted(unmatched_dropoff_ids)[:20])
```

## Retrieval Mode 2 — SQL retrieval

We'll use DuckDB because it can query Parquet and CSV directly without requiring us to build a complicated database first.

Parquet + CSV
     │
     ▼
   DuckDB
     │
     ▼
    SQL
     │
     ├── count trips
     ├── filter July 2026
     └── inspect zone relationship

```python
import duckdb

con = duckdb.connect()
```

```python
sql_trip_count = con.execute("""
    SELECT COUNT(*) AS trip_count
    FROM read_parquet('../data/raw/trips/yellow_tripdata_2026-07.parquet')
""").fetchdf()

print(sql_trip_count)
```

```python
sql_july = con.execute("""
    SELECT *
    FROM read_parquet('../data/raw/trips/yellow_tripdata_2026-07.parquet')
    WHERE tpep_pickup_datetime >= '2026-07-01'
      AND tpep_pickup_datetime < '2026-08-01'
""").fetchdf()

print("SQL July rows:", len(sql_july))
```

```python
print("Pandas July rows:", len(july_2026_trips))
print("SQL July rows:", len(sql_july))

print(
    "Counts match:",
    len(july_2026_trips) == len(sql_july)
)
```

```python
sql_zones = con.execute("""
    SELECT *
    FROM read_csv_auto('../data/raw/zones/taxi_zone_lookup.csv')
""").fetchdf()

print("SQL zone rows:", len(sql_zones))
print(sql_zones.head())
```

```python
print(
    "Zone counts match:",
    len(sql_zones) == len(zones)
)
```

```python
sql_zone_activity = con.execute("""
    SELECT
        t.PULocationID,
        z.Borough,
        z.Zone,
        COUNT(*) AS trip_count
    FROM read_parquet(
        '../data/raw/trips/yellow_tripdata_2026-07.parquet'
    ) t
    LEFT JOIN read_csv_auto(
        '../data/raw/zones/taxi_zone_lookup.csv'
    ) z
        ON t.PULocationID = z.LocationID
    WHERE t.tpep_pickup_datetime >= '2026-07-01'
      AND t.tpep_pickup_datetime < '2026-08-01'
    GROUP BY
        t.PULocationID,
        z.Borough,
        z.Zone
    ORDER BY trip_count DESC
    LIMIT 10
""").fetchdf()

print(sql_zone_activity)
```
